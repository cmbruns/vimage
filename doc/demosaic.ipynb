## Computing RGB in the corners

```python
from sympy import *

# Simplest case: 4 bayer cells, 1 shared corner; Linear/Nearest demosaic
# There is a single monochrome intensity in each cell.
I0, I1, I2, I3 = symbols("I_0, I_1, I_2, I_3")
bayer_block = Matrix([[I0, I1],[I2, I3]])
bayer_block
```

```python
# 1x1 block of corners/colors
C0 = symbols("C_0")
corner_block = Matrix([C0])
corner_block
```

```python
# There's really only one reasonable way to do it with this small neighborhood
# There are 4 kinds of corners: RGGB, GRBG, BGGR, GBRG
# Case 1: RGGB
red = I0
green = (I1 + I2) / 2
blue = I3
Eq(C0, Matrix([red, green, blue]), evaluate=False)
```
